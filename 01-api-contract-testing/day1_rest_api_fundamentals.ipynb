# Day 1 – REST API Fundamentals

## 1. What is an API?

API stands for Application Programming Interface.

An API allows two software applications to communicate with each other.

### Real-world example

Imagine an online shopping application.

1. You open the application.
2. You view your orders.
3. The application sends a request to the backend API.
4. The API retrieves order information.
5. The API sends the response back to the application.
6. You see your orders on the screen.

### API communication

Client → API Request → Server → API Response → Client

## 2. What is REST API?

REST stands for Representational State Transfer.

A REST API uses HTTP methods to allow applications to communicate with a server and access or modify resources.

Examples of resources:
- Customers
- Orders
- Products
- Payments



## 3. HTTP Methods

HTTP methods tell the server what action the client wants to perform.

| Method | Purpose | Example |
|---|---|---|
| GET | Retrieve data | View customer details |
| POST | Create new data | Register a customer |
| PUT | Replace existing data | Replace customer details |
| PATCH | Update selected fields | Change customer email |
| DELETE | Delete data | Delete a customer |

### Practical example

Imagine a customer management application.

GET /customers/101
- Retrieve customer 101.

POST /customers
- Create a new customer.

PUT /customers/101
- Replace the details of customer 101.

PATCH /customers/101
- Update only the email address.

DELETE /customers/101
- Delete customer 101.

### QA perspective

As a QA engineer, validate:

1. Does GET return the correct customer?
2. Does POST create a customer successfully?
3. Does PUT replace the expected details?
4. Does PATCH update only the requested fields?
5. Does DELETE remove the customer as expected?



## 4. API Request and Response Structure

An API request contains information sent by the client to the server.

An API response contains information returned by the server to the client.

### A. API Request

A request commonly contains:

1. HTTP Method – GET, POST, PUT, PATCH or DELETE.
2. Endpoint – The URL of the API.
3. Headers – Additional information about the request.
4. Request Body – Data sent to the server (when required).

### B. API Response

A response commonly contains:

1. Status Code – Indicates the result of the request.
2. Response Headers – Metadata about the response.
3. Response Body – Data returned by the server.
4. Response Time – How long the request took.

### Example

Request:

GET https://jsonplaceholder.typicode.com/users/1

Response:

{
  "id": 1,
  "name": "Leanne Graham",
  "username": "Bret",
  "email": "Sincere@april.biz"
}

Expected result:

- Status code: 200
- Customer ID: 1
- Name: Leanne Graham
- Email: Present

### QA perspective

A QA engineer should validate:

- Is the status code correct?
- Does the response contain the expected fields?
- Are the field values correct?
- Are data types correct?
- Is the response time acceptable?


```python

import requests

url = "https://jsonplaceholder.typicode.com/users/1"

response = requests.get(url)

print("Status Code:", response.status_code)
print("Response Body:")
print(response.json())
```

```python

# QA Validation

data = response.json()

assert response.status_code == 200
assert data["id"] == 1
assert data["name"] == "Leanne Graham"
assert data["username"] == "Bret"
assert "email" in data

print("All API validations passed!")
```

```python

# Negative Testing - Incorrect Expected Value

assert data["username"] == "John"

print("Test Passed")
```

```python

# Validate nested JSON fields

assert data["address"]["city"] == "Gwenborough"
assert data["company"]["name"] == "Romaguera-Crona"
assert data["address"]["geo"]["lat"] == "-37.3159"

print("Nested JSON validations passed!")
```
